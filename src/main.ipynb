```python
import sys
import typing
from datetime import datetime
from pathlib import Path
import logging

import mlflow

if typing.TYPE_CHECKING:
    from os import PathLike
```

```python
mlflow.set_experiment(experiment_id="0")
mlflow.pytorch.autolog()
```

```python
import setup
import files
import train
```

```python
_handler = logging.StreamHandler(sys.stdout)
_handler.setFormatter(
    logging.Formatter(
        "%(asctime)s - %(name)s - %(levelname)s - %(message)s", datefmt="%H:%M:%S'"
    )
)
_logger = logging.getLogger("adaptive_milling_training")
for _ in _logger.handlers:
    _logger.removeHandler(_)
_logger.setLevel(logging.DEBUG)
_logger.addHandler(_handler)
```

```python
def main(
    csv_path: str | PathLike[str], epochs: int = 30, cpu_only: bool = False
) -> None:
    csv_path = Path(csv_path).absolute()
    if not csv_path.is_file():
        raise FileNotFoundError(csv_path)

    _logger.info("Starting training with '%s'", csv_path)

    label_count = 5

    array = files.paths_array_from_csv(csv_path)

    training_data, validation_data = setup.create_datasets(
        *setup.split_array(array),
        image_size=1536,
        label_count=label_count,
        validation_split=0.2,
    )
    _logger.info("Datasets loaded")

    models_dir = Path.cwd().parent / "models"
    models_dir.mkdir(exist_ok=True)

    model_path = (
        models_dir / f"{datetime.now().strftime('%y%m%d_%H%M%S')}_{csv_path.stem}.pth"
    )
    _logger.info("The model will be saved as '%s'", model_path)

    with mlflow.start_run(description=model_path.stem):
        _logger.info("Starting training...")
        train.train(
            training_data,
            validation_data,
            label_count=label_count,
            model_path=model_path,
            epochs=epochs,
            cpu_only=cpu_only,
        )
```

```python
test_csv = "/ceph/groups/structbio/adaptive_milling_project/2024labels_new/test.csv"
all_files_csv = (
    "/ceph/groups/structbio/adaptive_milling_project/2024labels_new/all_files.csv"
)
main(all_files_csv, epochs=100)
```
